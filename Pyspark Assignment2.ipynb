```python
from pyspark.sql import SparkSession
```

```python
spark = (SparkSession.builder
            .appName("Assignment2")
            .master("local[2]")
            .getOrCreate())
```

```python
sc = spark.sparkContext
```

```python
rdd1 = sc.parallelize([3, 6, 9, 12, 15],2)
print(rdd1.collect())
print(rdd1.getNumPartitions())
print(rdd1.glom().collect())
```

```python
div = rdd1.map(lambda x: x//3)
print(div.collect())
```

```python
rdd2 = sc.parallelize(["a b","b c", "c a"])
print(rdd2.collect())
print(rdd2.getNumPartitions())
print(rdd2.glom().collect())
```

```python
Split = rdd2.flatMap(lambda x: x.split(" "))
pairs = Split.map(lambda x: (x, 1))
counts = pairs.reduceByKey(lambda x, y: x + y)
print(counts.collect())
```

```python
rdd3 = sc.parallelize(["Anu", "Asha","Ravi","Kiran"])
print(rdd3.collect())
print(rdd3.getNumPartitions())
print(rdd3.glom().collect())
```

```python
Filter = rdd3.filter(lambda x: x.startswith("A"))
print(Filter.collect())
```

```python
sales = sc.parallelize([("Chennai", 100),("Mumbai", 220),("Bangalore", 210),("Chennai", 130),
                        ("Mumbai", 300),("Bangalore", 350),("Chennai", 250),("Hyderabad", 300)],2)
print(sales.collect())
print(sales.getNumPartitions())
print(sales.glom().collect())
```

```python
combined_sales = sales.reduceByKey(lambda x, y: x + y)
high_sales = combined_sales.filter(lambda x: x[1] > 500)
print(combined_sales.collect())
print(high_sales.collect())
```

```python
total_sales = sales.map(lambda x: x[1]).sum()
print(total_sales)
```
